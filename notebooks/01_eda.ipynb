```python
import pandas as pd
from pathlib import Path 
```

```python
data_dir = Path("../data/raw")
```

```python
files = [
    "clicks_train.csv.zip",
    "events.csv.zip",
    "promoted_content.csv.zip",
    "documents_meta.csv.zip",
    "documents_categories.csv.zip",
    "documents_topics.csv.zip",
]

for file in files:
    df = pd.read_csv(data_dir / file)
    missing_ratio = df.isna().mean().round(3)
    print(file, df.shape)
    print(df.dtypes)
    print(missing_ratio[missing_ratio > 0])
```

```python
events = pd.read_csv(data_dir / "events.csv.zip", dtype={"platform": str})
events["platform"].value_counts(dropna=False)
```

```python
clicks_train = pd.read_csv(data_dir / "clicks_train.csv.zip")

candidates_per_display = clicks_train.groupby("display_id").size()
candidates_per_display.describe()
```

```python
clicks_train["clicked"].value_counts(normalize=True)
```

```python
documents_categories = pd.read_csv(data_dir / "documents_categories.csv.zip")
documents_topics = pd.read_csv(data_dir / "documents_topics.csv.zip")

rows_per_doc_category = documents_categories.groupby("document_id").size()
rows_per_doc_category.describe()
```

```python
rows_per_doc_topic = documents_topics.groupby("document_id").size()
rows_per_doc_topic.describe()
```

```python
documents_categories["confidence_level"].describe()
```

```python
documents_topics["confidence_level"].describe()
```

```python
promoted_content = pd.read_csv(data_dir / "promoted_content.csv.zip")

clicks_with_doc = clicks_train.merge(promoted_content[["ad_id", "document_id"]], on="ad_id", how="left")
doc_ids_in_clicks = clicks_with_doc["document_id"].dropna().unique()

pd.Series(doc_ids_in_clicks).isin(documents_categories["document_id"]).mean()
```
